💡 Lasso Regression (Least Absolute Shrinkage and Selection Operator) is a regularized linear regression technique that applies an L1 penalty to the model’s coefficients. This penalty reduces the impact of less important features by shrinking their coefficients toward zero, effectively performing feature selection.
📊 In this project, the dataset contains two columns — Experience and Salary. The goal is to predict an employee’s Salary based on their Years of Experience. By using Lasso Regression, we can create a model that not only fits the data well but also avoids overfitting by keeping the coefficients as simple as possible.
⚙️ Lasso helps identify how strongly Experience influences Salary, while maintaining a clean and interpretable model.
### Step 1: Import Libraries

```python
import pandas as pd
```

### Step 2: Import Dataset

Explanation:
We’re using the Salary dataset again, where:

X (features): Experience Years

y (target): Salary

```python
salary = pd.read_csv('dataset.csv')
salary.head()
```

### Step 3: Define Features and Target

X: Independent variable -> Experience Years

y: Dependent variable -> Salary

```python
X = salary[['Experience Years']]
y = salary['Salary']
```

### Step 4: Split Data into Training and Testing Sets

**train_test_split** -> to split data into training and testing parts

We split the data so that:

70% of it is used to train the model

30% is used to test it

**random_state** ensures the same split every time you run it

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=42)
```

### Step 5: Create Lasso Regression Model

Lasso Regression uses L1 regularization, which can shrink some coefficients exactly to zero, effectively performing feature selection.

The alpha parameter controls the regularization strength:

Higher alpha -> more penalty (simpler model)

Lower alpha -> less penalty (closer to Linear Regression)

```python
from sklearn.linear_model import Lasso
model = Lasso(alpha=0.1)
model.fit(X_train, y_train)
```

### Step 6: Check Model Parameters

Intercept: Predicted salary when experience = 0

Coefficient: Change in salary for each additional year of experience

```python
print("Intercept: ", model.intercept_)
print("Coefficient: ", model.coef_)
```

### Step 7: Make Predictions

We predict the salary for test data using the trained Lasso model.

```python
y_pred = model.predict(X_test)
y_pred
```

### Step 8: Evaluate Model Performance

We use three metrics to check performance:

MAE: Average prediction error

MSE: Penalizes large errors more

R² Score: Measures model accuracy (closer to 1 = better)

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("Mean Absolute Error: ", round(mean_absolute_error(y_test, y_pred), 3))
print("Mean Squared Error: ", round(mean_squared_error(y_test, y_pred), 3))
print("r2_Score: ", round(r2_score(y_test, y_pred), 3))
print("Accuracy: ", round(r2_score(y_test, y_pred)*100, 3),"%")
```
