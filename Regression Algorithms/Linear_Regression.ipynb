### Step 1 : import library
##### We import the pandas library because it helps us work easily with data in table format (called DataFrames).

```python
import pandas as pd
```

### Step 2 : import data
We load the Salary dataset directly from a URL using pd.read_csv().

This dataset contains two columns:

Experience Years → number of years a person has worked

Salary → their corresponding salary


```python
salary = pd.read_csv('dataset.csv')
```

### Step 3 : define target (y) and features (X)
X (features/input): The column used to make predictions → Experience Years

y (target/output): The value we want to predict → Salary

```python
salary.columns
y = salary['Salary']
X = salary[['Experience Years']]
```

### Step 4 : train test split
We divide the dataset into:

70% Training data → used to train the model

30% Testing data → used to test how well the model performs

random_state ensures we get the same random split every time (for consistency).
.shape shows how many rows are in each split.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, train_size=0.7, random_state=42)
```

```python
X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

### Step 5 : select model

We import the LinearRegression model from sklearn, which helps find the best-fitting straight line through the data.

Then we create an object called model that will be trained to learn the relationship between Experience and Salary.

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

### Step 6 : train or fit model

model.fit() → teaches the model using the training data.
It tries to find the best line y = m*x + b that fits the training points.

intercept_ → gives the value of b (the line’s starting point).

coef_ → gives the value of m (the slope, which tells how much salary increases per year of experience).

```python
model.fit(X_train,y_train)
model.intercept_
model.coef_
```

### Step 7 : predict model

Now we use the trained model to predict salaries for the test data.

model.predict() uses the formula it learned to calculate predicted salaries based on the experience values in X_test.

```python
y_pred = model.predict(X_test)
y_pred
```

### Step 8 : model accuracy

Mean Absolute Error (MAE): Average of how far predictions are from real values.

Mean Absolute Percentage Error (MAPE): Same as MAE, but in percentage form.

Mean Squared Error (MSE): Like MAE, but squares the errors (so big errors matter more).

R² Score: Measures how well our line fits the data.

Closer to 1 → better model

Closer to 0 → weaker fit

#### Higher R² means your model predicts salary accurately based on experience.

```python
from sklearn.metrics import mean_absolute_error, mean_absolute_percentage_error, mean_squared_error, r2_score, accuracy_score
print("mean_absolute_error:  ",round(mean_absolute_error(y_test,y_pred),3))
print("mean_absolute_percentage_error: ",round(mean_absolute_percentage_error(y_test,y_pred),3))
print("mean_squared_error:   ",round(mean_squared_error(y_test,y_pred),3))
print("r2_score:          ",round(r2_score(y_test,y_pred),2))
print("Accuracy:         ",round(r2_score(y_test,y_pred),2)*100,"%")
```
