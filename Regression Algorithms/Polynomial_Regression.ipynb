### Step 1: Import Libraries

pandas for data handling

numpy for numerical computing

```python
import pandas as pd
import numpy as np
```

### Step 2: Import Dataset

We load a simple Salary dataset where we predict Salary based on Experience Years.

```python
salary = pd.read_csv('https://github.com/ybifoundation/Dataset/raw/main/Salary%20Data.csv')
salary.head()
```

### Step 3: Define Features and Target

X → Independent variable (Experience)

y → Dependent variable (Salary)

```python
X = salary[['Experience Years']]
y = salary['Salary']
```

### Step 4: Train-Test Split

**train_test_split** for dividing data

We split data into:

70% training data

30% testing data

random_state ensures reproducibility



```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.7, random_state=2529)
```

### Step 5: Create Polynomial Features

**PolynomialFeatures** to create polynomial terms

PolynomialFeatures(degree=2) converts a single feature 
𝑥
x into 
[
1
,
𝑥
,
𝑥
2
]
[1,x,x
2
].

This allows the model to fit a curved line instead of a straight one.

```python
from sklearn.preprocessing import PolynomialFeatures
poly = PolynomialFeatures(degree=2)
X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)
```

### Step 6: Train Polynomial Regression Model

**LinearRegression** as the core model

We train a Linear Regression model on the polynomial-transformed features, effectively creating a Polynomial Regression.

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(X_train_poly, y_train)
```

### Step 7: Predict on Test Data

We predict salaries for test data using the trained model.

```python
y_pred = model.predict(X_test_poly)
print(y_pred)
```

### Step 8: Evaluate Model Performance

**metrics** for evaluating performance

MAE: Average absolute difference between predicted and actual salaries.

MSE: Measures squared error (penalizes big mistakes).

R² Score: Model fit quality (closer to 1 = better).

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print("Mean Absolute Error: ", round(mean_absolute_error(y_test, y_pred), 3))
print("Mean Squared Error: ", round(mean_squared_error(y_test, y_pred), 3))
print("r2_Score: ", round(r2_score(y_test, y_pred), 3))
print("Accuracy: ", round(r2_score(y_test, y_pred)*100, 3),"%")
```
